## 00.5 — Integração DuckDB, Python, Pandas e Arrow
DuckDB executa consultas analíticas SQL; Python organiza o fluxo; Pandas/Arrow ajudam no intercâmbio tabular.

- `duckdb.connect(':memory:')`: cria uma conexão com banco **não persistente**.
- `con.execute(sql).df()`: busca o resultado como **Pandas DataFrame**.
- `con.execute(sql).arrow()`: fornece resultado por interface **Apache Arrow** (com PyArrow instalado).
- **Atenção:** converter toda uma tabela enorme para Pandas pode consumir muita RAM; prefira limitar/filtrar/agregar antes da conversão.
- DuckDB pode ler Parquet sem `CREATE TABLE`; isso não implica “zero cópias” ou ausência de I/O em todos os cenários.

````python
# Diagnóstico explícito de dependências sem instalar nada
from importlib.util import find_spec
for pacote in ["duckdb", "pandas", "pyarrow", "ipykernel"]:
    print(f"{pacote:10s}:", "disponível" if find_spec(pacote) else "NÃO instalado")
````

````python
# Conexão DuckDB: execute apenas com a biblioteca instalada
import duckdb
import pandas as pd
con = duckdb.connect(':memory:')
print("DuckDB:", duckdb.__version__)
print(con.execute("SELECT 2 + 3 AS resultado").df())
````

## 00.8 — Troubleshooting (erros frequentes)
| Sintoma | Diagnóstico | Ação |
|---|---|---|
| Pacote instalado, import falha | Instalação em outro Python | Compare `sys.executable` e o kernel selecionado |
| Tabela sumiu | Nova conexão `:memory:` | Execute novamente o setup; evite criar conexões repetidas |
| Resultado antigo | Células fora de ordem | Reinicie e execute todas as células |
| Caminho de arquivo não funciona | Diretório atual diferente | Verifique `Path.cwd()` e caminhos relativos |
| `assert` falhou | Invariante não respeitada | Inspecione a consulta que encontrou registros problemáticos |

**Atenção:** `assert` auxilia no laboratório; em pipelines produtivos use validação explícita, pois `python -O` pode desativar asserts.

## 00.10 — Primeiro laboratório DuckDB: SLA operacional
**Cenário:** três equipes de manutenção registraram chamados concluídos. Queremos total por equipe e validar o total geral.

**Hipótese:** Centro possui 35 chamados, Norte 12 e Sul 17. O total deve ser 64.

**Tarefa:** execute a próxima célula, compare o resultado e depois modifique um valor para prever como o total se altera.

````python
# LAB 00.A — SQL executado pelo DuckDB
con.execute("""
    CREATE OR REPLACE TABLE chamados AS
    SELECT * FROM (VALUES
        ('Centro', 20), ('Centro', 15),
        ('Norte', 12), ('Sul', 8), ('Sul', 9)
    ) AS t(equipe, concluidos);
""")
resumo = con.execute("""
    SELECT equipe, SUM(concluidos) AS total_concluidos
    FROM chamados
    GROUP BY equipe
    ORDER BY equipe
""").df()
display(resumo)
````

````python
# LAB 00.B — Verificação automática da regra de negócio
totais = dict(zip(resumo['equipe'], resumo['total_concluidos']))
assert totais == {'Centro': 35, 'Norte': 12, 'Sul': 17}, f'Totais inesperados: {totais}'
assert sum(totais.values()) == 64, 'Total geral divergente'
print('✅ Dados validados: 64 chamados concluídos.')
````
