Nesta etapa, carregamos as bibliotecas essenciais (`duckdb`, `pandas`), instalamos o driver de conexão e configuramos o ambiente para aceitar comandos SQL nativos.

````python
import duckdb
import pandas as pd

# 1. Instalação e Carregamento da Extensão
!pip install duckdb-engine --quiet
%reload_ext sql

# 2. Configurações Globais do SQL Magic
%config SqlMagic.autopandas = True
%config SqlMagic.feedback = False
%config SqlMagic.displaycon = False
%config SqlMagic.style = 'plain'

# 3. Conexão ao banco de dados DuckDB
con = duckdb.connect(database=':memory:')

# Variável global para armazenar o último resultado
last_result = None

# 4. Função Auxiliar para Simplificar o SQL
def sql(query):
    """Executa uma query SQL, exibe e armazena o resultado"""
    global last_result
    last_result = con.execute(query).df()
    display(last_result)
    # Removi o 'return' para evitar a exibição duplicada no Colab
````

Nesta etapa, o arquivo CSV é carregado via Pandas e imediatamente registrado no catálogo do DuckDB. Isso transforma o arquivo bruto em uma **tabela SQL relacional** de alta performance, pronta para consulta.

### 🔗 Etapa 2: Registro no Catálogo SQL
Para que o motor SQL reconheça seu DataFrame como uma tabela, precisamos registrá-lo. No código abaixo, você define o **nome da tabela** que será usada em suas consultas `SELECT`.

````python
# Registro oficial no DuckDB
nome_da_tabela = 'baby_names'
con.register(nome_da_tabela, df_input)

print(f"Pronto! A tabela '{nome_da_tabela}' está disponível para consultas SQL.")
sql(f"SELECT * FROM {nome_da_tabela} LIMIT 5")
````

## 🧠 Explorando o Potencial do DuckDB

O **DuckDB** é frequentemente chamado de "SQLite para Analytics". Aqui estão as respostas para suas dúvidas de integração:

### 1. Conexão com Power BI
Como o DuckDB no Colab roda temporariamente na memória da nuvem do Google, o Power BI não consegue "enxergar" esse arquivo diretamente via rede.
*   **A Solução Corporativa:** Geralmente, usamos o DuckDB para processar grandes volumes de dados (ETL) e salvamos o resultado em arquivos **Parquet**. O Power BI lê arquivos Parquet de forma extremamente rápida, sendo esta a melhor forma de integrar as duas ferramentas.

### 2. Desenvolvimento de Aplicações
Sim! O DuckDB possui bibliotecas para **Java, C++, Rust, Node.js e Go**. Se você criar um software (como um Dashboard em Streamlit ou uma API), o DuckDB pode ser o motor interno de processamento de dados, oferecendo performance de Data Warehouse dentro do seu app.

### 3. Principais Diferenciais
*   **Processamento Vetorial:** Ele processa colunas inteiras de uma vez, não linha por linha, o que o torna centenas de vezes mais rápido que bancos tradicionais para somas e médias.
*   **Portabilidade:** O banco pode ser salvo em um único arquivo `.db` no seu Google Drive, mantendo todos os seus dados e tabelas persistentes.

## Reinicializar Motor SQL e Dependências

### Subtask:
Configurar o motor analítico DuckDB e definir a função auxiliar para execução de consultas SQL.


**Reasoning**:
I will initialize the DuckDB connection, define the sql helper function, and set up the global variable for result tracking as per the subtask instructions.



````python
import duckdb
import pandas as pd

# 1. Create in-memory DuckDB connection
con = duckdb.connect(database=':memory:')

# 2. Initialize global variable for tracking last result
last_result = None

# 3. Define helper function for SQL execution
def sql(query):
    """Executes a SQL query, displays, and stores the result."""
    global last_result
    try:
        last_result = con.execute(query).df()
        display(last_result)
    except Exception as e:
        print(f"Error executing query: {e}")

print("DuckDB engine initialized and sql() function ready.")
````
