## 2. 🪄 Magia Negra do IPython: Comandos Mágicos de Linha (%) e Bloco (%%)

Comandos que iniciam com `%` afetam uma única linha; comandos com `%%` afetam todo o conteúdo da célula.

````python
# 1. Benchmark de Precisão: %time vs %timeit
import numpy as np

# %time mede a execução única (inclui latência de inicialização)
%time lista = [x**2 for x in range(1000000)]

# %timeit executa múltiplas repetições para obter média e desvio padrão reais
%timeit arr = np.arange(1000000)**2
````

````python
%%writefile modulo_exemplo.py
# %%writefile grava todo o conteúdo desta célula diretamente em um arquivo no disco da VM!
def calcular_indicador_telecom(capacidade_mbps, consumo_medio):
    taxa_ocupacao = (consumo_medio / capacidade_mbps) * 100
    return round(taxa_ocupacao, 2)

print('Módulo salvo no disco da VM!')
````

````python
# Importando o arquivo recém-criado na VM sem precisar de upload manual
import modulo_exemplo as me
print(f"Taxa de Ocupação: {me.calcular_indicador_telecom(1000, 785)}%")
````

## 6. 🛡️ Prevenção de Desconexões e Gerenciamento de Memória (Anti-OOM)

### Políticas de Desconexão do Google Colab Gratuito
1. **Timeout por Inatividade (90 minutos):** se você fechar o navegador ou não interagir com o notebook por 90 minutos, a sessão é desconectada.
2. **Timeout Absoluto (12 horas):** qualquer sessão tem tempo máximo contínuo de 12 horas, após o qual a VM é reciclada.

### Hack de Anti-Inatividade no Navegador
Se você estiver executando uma query de processamento pesado ou treinamento longo, abra o console do desenvolvedor do navegador (`F12` ou `Ctrl + Shift + I` > aba **Console**) e execute este script JavaScript:

```javascript
function manterConectado() {
    console.log("Mantendo sessão do Colab ativa...");
    document.querySelector("#top-toolbar > colab-connect-button").shadowRoot.querySelector("#connect").click();
}
setInterval(manterConectado, 60000); // Executa a cada 60 segundos
```

### Como Liberar Memória RAM Sem Reiniciar a Máquina

````python
import gc

# Deleta variáveis gigantes que não serão mais usadas
try:
    del df_demo
except NameError:
    pass

# Força a coleta de lixo do Python para devolver a memória ao sistema operacional
gc.collect()
print('Memória RAM liberada com sucesso!')
````

## 7. 📱 Experiência Multi-Dispositivo: Computador, Tablet e Celular

O Google Colab é uma aplicação web 100% responsiva hospedada na nuvem. Você pode estudar, revisar e executar códigos analíticos em qualquer lugar, alternando perfeitamente entre seu computador de trabalho, seu tablet no sofá ou seu smartphone em deslocamento.

### 🖥️ 1. No Computador (Desktop / Notebook)
- **Produtividade Máxima:** uso intensivo dos atalhos de teclado (`Shift + Enter`, `Ctrl + M B`, `Ctrl + Shift + P`).
- **Painel Lateral de Variáveis:** clique no ícone `{x}` na barra lateral esquerda para inspecionar todas as variáveis, DataFrames e tipos carregados em memória em tempo real.
- **Visualização Ampla:** capacidade de exibir matrizes largas com dezenas de colunas simultâneas.

### 📟 2. No Tablet (iPad / Galaxy Tab)
- **Modo Split-Screen (Tela Dividida):** a configuração mais produtiva para estudantes e instrutores. Abra o navegador com duas janelas lado a lado — o [Livro no Google Docs] à esquerda e o [Google Colab] à direita para ler a teoria e rodar o código simultaneamente.
- **Teclados Bluetooth:** ao conectar um teclado físico (Magic Keyboard ou teclado Bluetooth), todos os atalhos de desktop funcionam nativamente (`Cmd` no iPad e `Ctrl` no Android).

### 📱 3. No Celular / Smartphone (Android & iPhone)
- **Botão de Play com Um Toque:** cada célula possui um botão circular de play grande, permitindo rodar queries pesadas com apenas um toque na tela sem precisar digitar.
- **Modo de Exibição 'Versão para Computador':** se a barra de ferramentas parecer restrita no navegador mobile, toque no menu do navegador (três pontinhos no Chrome ou botão `aA` no Safari) e ative **'Para Computador' / 'Solicitar Site para Desktop'** para liberar todos os menus de topo.

### ⚙️ Configuração de Exibição Responsiva (Telas Estreitas)
Ao rodar no celular ou tablet, configure as bibliotecas para não quebrar tabelas na tela vertical:

````python
# Configuração de Visualização Responsiva para Telas Pequenas / Celular
import polars as pl
import pandas as pd

# 1. Polars: Limita largura e colunas para caber em telas verticais de celular
pl.Config.set_tbl_cols(4)           # Exibe até 4 colunas principais
pl.Config.set_tbl_width_chars(60)   # Ajusta largura total em caracteres

# 2. Pandas: Limita colunas visíveis
pd.set_option('display.max_columns', 4)
pd.set_option('display.width', 60)

print('Ambiente configurado com sucesso para visualização móvel e telas compactas!')
````
