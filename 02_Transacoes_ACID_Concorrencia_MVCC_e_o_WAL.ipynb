## 02.4 — Laboratório guiado

**Previsão:** antes de executar, descreva em uma frase o resultado esperado.

**Interpretação:** depois de executar, explique o que cada coluna ou efeito representa.

```python
con.execute("CREATE TABLE saldos (id INTEGER, saldo INTEGER)")
con.execute("INSERT INTO saldos VALUES (1,100),(2,100)")
con.execute('BEGIN TRANSACTION')
con.execute('UPDATE saldos SET saldo = saldo - 30 WHERE id = 1')
con.execute('UPDATE saldos SET saldo = saldo + 30 WHERE id = 2')
con.execute('ROLLBACK')
display(con.execute('SELECT * FROM saldos ORDER BY id').df())
assert con.execute('SELECT SUM(saldo) FROM saldos').fetchone()[0] == 200
```

## 02.5 — Experimentação controlada

Altere um parâmetro de cada vez. Compare a saída com a hipótese e registre se a mudança alterou os dados, o plano ou apenas a apresentação.

```python
con.execute('BEGIN TRANSACTION')
con.execute('UPDATE saldos SET saldo = saldo - 30 WHERE id = 1')
con.execute('UPDATE saldos SET saldo = saldo + 30 WHERE id = 2')
con.execute('COMMIT')
assert con.execute('SELECT saldo FROM saldos WHERE id=1').fetchone()[0] == 70
display(con.execute('SELECT * FROM saldos ORDER BY id').df())
```
